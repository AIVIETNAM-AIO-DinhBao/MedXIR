```python
import resource, traceback
def rss():
    return round(resource.getrusage(resource.RUSAGE_SELF).ru_maxrss / 2**20, 2)
def run(name, fn):
    print('START', name, 'rss GiB', rss(), flush=True)
    try:
        fn()
        print('DONE', name, 'rss GiB', rss(), flush=True)
    except Exception:
        print('FAILED', name, traceback.format_exc(), flush=True)
```

```python
import glob, numpy as np, torch
path = glob.glob('/kaggle/input/**/vectors/en-part-000.npy', recursive=True)[0]
print('file', path, flush=True)
def read_full():
    global arr
    arr = np.load(path)
    print('loaded', arr.shape, arr.dtype, flush=True)
run('np.load full vector part', read_full)
def to_gpu():
    v = torch.from_numpy(arr).cuda().float()
    print('gpu', tuple(v.shape), torch.cuda.memory_allocated() / 2**30, 'GiB', flush=True)
run('copy to GPU as float32', to_gpu)
print('probe complete', flush=True)
```
