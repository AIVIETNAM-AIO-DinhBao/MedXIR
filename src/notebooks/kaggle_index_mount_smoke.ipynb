```python
import glob, json, os
from pathlib import Path
roots = sorted(glob.glob('/kaggle/input/**/index_manifest.json', recursive=True))
assert len(roots) == 1, roots
root = Path(roots[0]).parent
files = sorted(root.glob('*.faiss'))
assert len(files) == 9, [(p.name, p.stat().st_size) for p in files]
result = {'root': str(root), 'index_count': len(files),
          'bytes': sum(p.stat().st_size for p in files)}
print(json.dumps(result))
Path('/kaggle/working/index_mount.json').write_text(json.dumps(result))
```
